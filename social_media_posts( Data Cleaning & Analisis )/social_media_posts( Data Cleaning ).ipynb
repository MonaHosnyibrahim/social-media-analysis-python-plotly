# Data Analysis

```python
reach=df[['Likes','Shares','Comments']]
```

```python
reach.describe().astype(int)
```

```python
df['Reach']=reach.sum(axis=1)
df
```

```python
#post type by Reach
df.groupby('Post_Type')['Reach'].mean().astype(int).sort_values(ascending=False)
```

```python
bar_chart = df.groupby('Post_Type', as_index=False)['Reach'].mean().astype({"Reach": int})

fig1 = px.bar(
    bar_chart,
    x="Post_Type",
    y="Reach",
    title="Post Type Average Reach",
    color="Post_Type",
    color_discrete_sequence=px.colors.sequential.Purples,
    text="Reach"
)

fig1.show()
```

```python
#city by Reach
df.groupby('City')['Reach'].mean().astype(int).sort_values(ascending=False)
```

```python
#most year got reach
df.groupby('Year')['Reach'].mean().astype(int).sort_values(ascending=False)
```

```python
lin_plot = df.groupby("Year", as_index=False)["Reach"].mean()

fig = px.line(
    lin_plot,
    x="Year",
    y="Reach",
    markers=True,
    title="Average Reach by Year"
)
fig.update_xaxes(
    tickmode="array",
    tickvals=[2024, 2025]
)

fig.show()
```

```python
corr=reach.corr()
fig = px.imshow(
    corr,
    text_auto=".2f",
    color_continuous_scale="RdBu",
    zmin=-1,
    zmax=1,
    title="Correlation Heatmap"
)

fig.show()
```
